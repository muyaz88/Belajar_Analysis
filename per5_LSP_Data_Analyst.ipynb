```python
f_c.fillna({'country': 'kosong', 'rating': 'kosong', 'duration': 'kosong'},inplace=True)
```

```python
f_c = f_c.dropna(how="any", subset=['director'])
```

```python
f_c = f_c.dropna(how="any", subset=['cast'])
```

```python
plt.figure(figsize = (15,15))
plt.boxplot(f_c["release_year"])
```

```python
f_c["release_year"].describe()
```

```python
Q1 = f_c["release_year"].quantile(0.25)
Q3 = f_c["release_year"].quantile(0.75)
IQR = Q3 - Q1
```

```python
outlierQ3 = Q3 + 1.5 * IQR
```

```python
f_c = f_c[(f_c["release_year"] < outlierQ3)]
f_c["release_year"].describe()
```

```python
f_c.to_csv('netflixclean.csv', index=False)
```
