# Impor data dari CSV ke DataFrame

```python
df = pd.read_csv('Online Retail Data.csv', header=0)
#mengasih batasan untuk data yang digunakan
df = df[df['price'].between(0, 100, inclusive='right')].reset_index(drop=True)
df
```

```python
df.info()
```

# Membuat data binning di DataFrame

## Binning dengan banyaknya bins tertentu dan panjang masing-masing bin sama

```python
#pengelompokan dari 3 bin berdasarkan harga mengguanakan pandas otomatis
df1 = df.copy()
category = ['Cheap','Medium','Expensive']
df1['bin1'] = pd.cut(df1['price'], 3, labels=category, include_lowest=True)
df1
```

```python
# mencari tahu pengolongannya
bin1_summary = df1.groupby('bin1').agg(row_cnt=('product_code','count'), min_price=('price','min'), max_price=('price','max'))
bin1_summary['bin_interval'] = bin1_summary['max_price'] - bin1_summary['min_price']
bin1_summary
```

```python
# menentukan 4 interval berdasarkan np.linspace
df2 = df1.copy()
bins = np.linspace(min(df['price']), max(df['price']), 4)
category = ['Cheap','Medium','Expensive']
df2['bin2'] = pd.cut(df2['price'], bins, labels=category, include_lowest=True)
df2
```

```python
bins
```

```python
# mencari tahu pengolongannya
bin2_summary = df2.groupby('bin2').agg(row_cnt=('product_code','count'), min_price=('price','min'), max_price=('price','max'))
bin2_summary['bin_interval'] = bin2_summary['max_price'] - bin2_summary['min_price']
bin2_summary
```

## Binning dengan panjang masing-masing interval tertentu

```python
# menentukan intervalnya berdasarkan tentuan min max harganya dengan pembagian 10 frequensi
df3 = df2.copy()
interval_range = pd.interval_range(start=min(df['price']), freq=10, end=max(df['price']))
df3['bin3'] = pd.cut(df['price'], bins=interval_range, include_lowest=True)
df3
```

```python
# melihat interval rangenya
interval_range
```

```python
# mencari tahu pengolongannya yang telah di tentukan dari freq sebanyak 10 sehingga binsnya menarik otomatis intervalnya sebanyak 10 gap, dilihat dari max pricenya
bin3_summary = df3.groupby('bin3').agg(row_cnt=('product_code','count'), min_price=('price','min'), max_price=('price','max'))
bin3_summary['bin_interval'] = bin3_summary['max_price'] - bin3_summary['min_price']
bin3_summary
```

## Binning dengan panjang masing-masing bin yang ditentukan sendiri sesuai kebutuhan

```python
# kita melakukan intervalnya sebanyak 3 dengan penentuan dari kita dengan pengcategoriannya sebanyak 3
df4 = df3.copy()
bins_edges = [0, 30, 80, 100]
category = ['Cheap','Medium','Expensive']
df4['bin4'] = pd.cut(df['price'], bins=bins_edges, labels=category, include_lowest=True)
df4
```

```python
# melihat jarak masing2 interval yang berbeda2 sesui dengan pendevinisian yang kita buat
bin4_summary = df4.groupby('bin4').agg(row_cnt=('product_code','count'), min_price=('price','min'), max_price=('price','max'))
bin4_summary['bin_interval'] = bin4_summary['max_price'] - bin4_summary['min_price']
bin4_summary
```

## Binning dengan banyaknya bins tertentu dan banyaknya anggota setiap bin sama

```python
# membuat beans dengan paling simple karna kita mendevinisikan hanya 3 kategori dengan panjang intervalnya di tentukan sistem
df5 = df4.copy()
category = ['Cheap','Medium','Expensive']
df5['bin5'] = pd.qcut(df['price'], q=3, labels=category)
df5
```

```python
# melihat summary bedasarkan intervalnya untuk melihat variasinya
bin5_summary = df5.groupby('bin5').agg(row_cnt=('product_code','count'), min_price=('price','min'), max_price=('price','max'))
bin5_summary['bin_interval'] = bin5_summary['max_price'] - bin5_summary['min_price']
bin5_summary
```

# Mengelompokkan data dengan case statement

```python
# melakukan pengelompokan berdasarkan kategiri produk yang telah kita tentukan menggunakan case statement
df6 = df5.copy()
df6['price_class'] = np.select(
    [(df6['product_name'].str.lower().str.contains('tea')) & (df6['price']<=2),
     (df6['product_name'].str.lower().str.contains('tea')) & (df6['price'].between(2, 4, inclusive='right')),
     (df6['product_name'].str.lower().str.contains('tea')) & (df6['price']>4),
     (df6['product_name'].str.lower().str.contains('coffee')) & (df6['price']<=3),
     (df6['product_name'].str.lower().str.contains('coffee')) & (df6['price'].between(3, 5, inclusive='right')),
     (df6['product_name'].str.lower().str.contains('coffee')) & (df6['price']>5)],
    ['Cheap Tea', 'Medium Tea', 'Expensive Tea', 'Cheap Coffee', 'Medium Coffee', 'Expensive Coffee'],
    default='Undetermined'
)
df6
```

```python
# melihat summarynya untuk melihat interval nya berdasarkan apa pengelompokkannya
price_class_summary = df6.groupby('price_class').agg(row_cnt=('product_code','count'), min_price=('price','min'), max_price=('price','max'))
price_class_summary['price_interval'] = price_class_summary['max_price'] - price_class_summary['min_price']
price_class_summary
```
