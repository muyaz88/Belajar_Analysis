```python
#Sebutkan 5 kategori yang memiliki rata-rata jumlah potongan harga terbesar. Sebutkan juga masing-masing nilai dari kategori tersebut!
pd.read_sql_query(""" SELECT category, AVG(discount_percentage) AS avg_discount
FROM salesamazon
GROUP BY category
ORDER BY avg_discount DESC
LIMIT 5;""",koneksi)
```

```python
#Berapa jumlah user yang memberikan review yang memuat kata 'Good Product' pada produk dengan kategori Computers & Accessories dan Office Products ?
pd.read_sql_query(""" SELECT distinct(category), count(review_title) as jumlah_review  FROM salesamazon
WHERE category IN ("Computers & Accessories","Office Products")
AND (review_title LIKE "%Good Product%" OR review_content LIKE "%Good Product%")
GROUP BY category
;""",koneksi)
```

# NO. 3

```python
#Tentukan 5 bulan yang memiliki total sales/revenue terendah pada tahun 2004 (dari seluruh kategori produk kecuali Office Products dan Home & Kitchen). Sebutkan juga nilai setiap revenue nya!
pd.read_sql_query(""" SELECT
category,
discounted_price*qty AS total_sales,
strftime('%m', order_date) AS bulan,
strftime('%Y', order_date) AS tahun

from salesamazon

WHERE order_date >= "2004-01-01"
AND order_date < "2005-01-01"
AND category NOT IN ("Office Products","Home & Kitchen")

GROUP BY bulan, category
ORDER BY total_sales ASC, category
LIMIT 5
;""",koneksi)
```

```python
#Sebutkan kategori apa yang rata-rata waktu pengirimannya paling cepat dan lambat (Tanpa kategori Car & Motorbike dan Toys & Games).
pd.read_sql_query("""
SELECT
  category,
  order_date,
  arrival_date,
  ROUND(avg(julianday(DATE(arrival_date)) - julianday(DATE(order_date)))) AS lama_pengiriman
FROM salesamazon
WHERE category NOT IN ('Car & Motorbike', 'Toys & Games')
GROUP BY category
ORDER BY lama_pengiriman ASC
;""",koneksi)
```

```python
#Hitunglah AOV pada setiap tahunnya, namun data yang digunakan haruslah data dari awal hingga pertengahan bulan (tanggal 01 - 15) untuk setiap bulannya
pd.read_sql_query("""
SELECT
  total_sales,
  total_order,
  avg(total_sales/total_order) AS aov,
  year
FROM
  (SELECT
    discounted_price * qty AS total_sales,
    COUNT(product_id) AS total_order,
    strftime('%Y', order_date) AS year,
    order_date
  FROM salesamazon
  WHERE order_date BETWEEN strftime('%Y-%m-01', order_date) AND strftime('%Y-%m-15', order_date)
  GROUP BY year) subquery

  group by year
;""",koneksi)
```
