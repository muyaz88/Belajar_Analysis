```python
#mengimport library yang akan digunakan
import sqlite3
import pandas as pd
```

```python
#membuat koneksi
koneksi = sqlite3.connect('iowa_sales.db')
```

```python
#membuat kursor
kursor = koneksi.cursor()
```

# NO 1,2

```python
#Membuat variabel bernama "CekTabelDatabase" untuk mempermudah mengecek tabel yang ada di database iowa_sales
CekTabelDatabase = """
SELECT name FROM sqlite_master WHERE type='table' AND name NOT LIKE 'sqlite_%';

"""
kursor.execute(CekTabelDatabase)
pd.read_sql_query(CekTabelDatabase,koneksi)
```

```python
#Menghapus tabel employee
query = """ DROP TABLE employee;

"""

kursor.execute(query)
```

```python
#Mengecek kembali tabel yang ada di database iowa_sales
pd.read_sql_query(CekTabelDatabase,koneksi)
```
